Vaccination is an important public health intervention for preventing and controlling infectious diseases. However, vaccination coverage, disease incidence, reported cases, and vaccine introduction timelines can vary across countries, regions, diseases, and years.

The challenge is to analyze these datasets together to understand **where vaccination coverage is high or low, how disease incidence and reported cases change over time, and how vaccination programs and vaccine introductions are distributed geographically**.

This project aims to transform raw global vaccination datasets into clean and structured data and use analytical techniques to identify meaningful trends and relationships.

The analysis focuses on:

- Understanding global vaccination coverage patterns
- Examining disease incidence and reported cases over time
- Comparing vaccination coverage across countries and regions
- Analyzing vaccine introduction timelines
- Exploring vaccination schedules and target populations
- Investigating relationships between vaccination coverage and disease indicators
- Identifying geographic and temporal patterns that can support public health analysis

The cleaned and analyzed data is ultimately used to build an **interactive Power BI dashboard** that enables users to explore vaccination and disease trends across countries, diseases, and years.

````python

# Convert the cleaned Coverage dataframe into MySQL INSERT statements.
#
# Why?
# The CSV importer is having trouble with commas and NULL values.
# Creating SQL statements lets us control exactly how every value
# is inserted into MySQL.

import pandas as pd
import numpy as np

def mysql_value(value):
    """
    Convert a Pandas value into a MySQL-safe SQL value.

    - NaN/None -> NULL
    - Text -> quoted and escaped
    - Numbers -> inserted as numbers
    """

    # Handle missing values
    if pd.isna(value):
        return "NULL"

    # Handle text values
    if isinstance(value, str):
        # Escape single quotes for MySQL
        value = value.replace("\\", "\\\\").replace("'", "''")
        return f"'{value}'"

    # Handle numeric values
    return str(value)


# Column names in the same order as the MySQL table
columns = [
    "`GROUP`",
    "CODE",
    "NAME",
    "YEAR",
    "ANTIGEN",
    "ANTIGEN_DESCRIPTION",
    "COVERAGE_CATEGORY",
    "COVERAGE_CATEGORY_DESCRIPTION",
    "TARGET_NUMBER",
    "DOSES",
    "COVERAGE"
]

# Create SQL file
with open("/content/coverage_import.sql", "w", encoding="utf-8") as f:

    f.write("USE vaccination_analysis;\n\n")

    # Insert data in batches of 1000 rows
    batch_size = 1000

    for start in range(0, len(coverage_df), batch_size):

        batch = coverage_df.iloc[start:start + batch_size]

        values = []

        for _, row in batch.iterrows():

            row_values = [
                mysql_value(row["GROUP"]),
                mysql_value(row["CODE"]),
                mysql_value(row["NAME"]),
                mysql_value(row["YEAR"]),
                mysql_value(row["ANTIGEN"]),
                mysql_value(row["ANTIGEN_DESCRIPTION"]),
                mysql_value(row["COVERAGE_CATEGORY"]),
                mysql_value(row["COVERAGE_CATEGORY_DESCRIPTION"]),
                mysql_value(row["TARGET_NUMBER"]),
                mysql_value(row["DOSES"]),
                mysql_value(row["COVERAGE"])
            ]

            values.append("(" + ",".join(row_values) + ")")

        f.write(
            "INSERT INTO coverage (" +
            ",".join(columns) +
            ") VALUES\n"
        )

        f.write(",\n".join(values))
        f.write(";\n\n")

print("SQL file created successfully.")
print("Rows:", len(coverage_df))
````

````python
# Create multiple smaller SQL files.
#
# Why?
# phpMyAdmin may stop processing a very large SQL file.
# Splitting the data into smaller files makes each import
# easier and more reliable.

import os

# Folder where the SQL files will be stored
output_folder = "/content/coverage_sql_parts"
os.makedirs(output_folder, exist_ok=True)

# Number of rows in each SQL file
chunk_size = 20000

# MySQL table columns
columns = [
    "`GROUP`",
    "CODE",
    "NAME",
    "YEAR",
    "ANTIGEN",
    "ANTIGEN_DESCRIPTION",
    "COVERAGE_CATEGORY",
    "COVERAGE_CATEGORY_DESCRIPTION",
    "TARGET_NUMBER",
    "DOSES",
    "COVERAGE"
]

def mysql_value(value):
    # Missing value -> SQL NULL
    if pd.isna(value):
        return "NULL"

    # Text value -> safely quote for MySQL
    if isinstance(value, str):
        value = value.replace("\\", "\\\\")
        value = value.replace("'", "''")
        return f"'{value}'"

    # Numeric value
    return str(value)


# Split dataframe into smaller files
file_number = 1

for start in range(0, len(coverage_df), chunk_size):

    chunk = coverage_df.iloc[start:start + chunk_size]

    file_path = f"{output_folder}/coverage_part_{file_number}.sql"

    with open(file_path, "w", encoding="utf-8") as f:

        f.write("USE vaccination_analysis;\n\n")

        values = []

        for _, row in chunk.iterrows():

            row_values = [
                mysql_value(row["GROUP"]),
                mysql_value(row["CODE"]),
                mysql_value(row["NAME"]),
                mysql_value(row["YEAR"]),
                mysql_value(row["ANTIGEN"]),
                mysql_value(row["ANTIGEN_DESCRIPTION"]),
                mysql_value(row["COVERAGE_CATEGORY"]),
                mysql_value(row["COVERAGE_CATEGORY_DESCRIPTION"]),
                mysql_value(row["TARGET_NUMBER"]),
                mysql_value(row["DOSES"]),
                mysql_value(row["COVERAGE"])
            ]

            values.append("(" + ",".join(row_values) + ")")

        f.write(
            "INSERT INTO coverage (" +
            ",".join(columns) +
            ") VALUES\n"
        )

        f.write(",\n".join(values))
        f.write(";")

    print(f"Created: coverage_part_{file_number}.sql | Rows: {len(chunk)}")

    file_number += 1

print("\nAll SQL parts created.")
````

### Why?

Exporting the cleaned DataFrames to CSV gives us a stable, portable

# **Exploratory Data Analysis (EDA)**

## Objective

With the datasets cleaned, this section explores vaccination coverage,
disease incidence, and vaccine introduction trends to answer several
of the project's guiding questions. Each sub-section states the
question, the approach, a chart, and an observation drawn from the
chart, in the same "Why? / Observation" style used during data
cleaning.

````python
# Set a consistent, readable style for all charts in this section.
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (11, 6)
plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.titleweight"] = "bold"
````

## Question 1: How has global vaccination coverage changed over time?

We look at the global (`WUENIC`/`ADMIN` estimate) average coverage
over time for two widely-tracked antigens: **DTPCV3** (3rd dose of the
diphtheria-tetanus-pertussis vaccine) and **MCV1** (1st dose of the
measles-containing vaccine).

````python
key_antigens = ["DTPCV3", "MCV1", "POL3", "BCG"]

coverage_trend = (
    coverage_df[coverage_df["ANTIGEN"].isin(key_antigens)]
    .groupby(["YEAR", "ANTIGEN"])["COVERAGE"]
    .mean()
    .reset_index()
)

plt.figure()
sns.lineplot(
    data=coverage_trend, x="YEAR", y="COVERAGE",
    hue="ANTIGEN", marker="o"
)
plt.title("Global Average Vaccination Coverage Over Time")
plt.xlabel("Year")
plt.ylabel("Average Coverage (%)")
plt.legend(title="Antigen")
plt.tight_layout()
plt.show()
````

### Observation

Global average coverage for **DTPCV3**, **MCV1**, **POL3**, and
**BCG** all rose substantially from the 1980s onward as the WHO
Expanded Programme on Immunization scaled up, before leveling off at
a high plateau in more recent years. Coverage generally stays below
100%, reflecting real-world delivery gaps rather than a data issue.

## Question 2: What is the drop-off rate between the 1st and later
vaccine doses?

We compare **DTPCV1** (1st dose) with **DTPCV3** (3rd dose) coverage
by year. The gap between the two lines is the "drop-off": people who
started but did not complete the recommended dose series.

````python
dtp = (
    coverage_df[coverage_df["ANTIGEN"].isin(["DTPCV1", "DTPCV3"])]
    .groupby(["YEAR", "ANTIGEN"])["COVERAGE"]
    .mean()
    .reset_index()
    .pivot(index="YEAR", columns="ANTIGEN", values="COVERAGE")
    .dropna()
)

dtp["DROPOFF_RATE_PCT"] = (
    (dtp["DTPCV1"] - dtp["DTPCV3"]) / dtp["DTPCV1"] * 100
)

fig, ax1 = plt.subplots()
ax1.plot(dtp.index, dtp["DTPCV1"], marker="o", label="DTP dose 1 coverage")
ax1.plot(dtp.index, dtp["DTPCV3"], marker="o", label="DTP dose 3 coverage")
ax1.set_xlabel("Year")
ax1.set_ylabel("Coverage (%)")
ax1.legend(loc="lower right")

ax2 = ax1.twinx()
ax2.bar(dtp.index, dtp["DROPOFF_RATE_PCT"], alpha=0.15, color="red", label="Drop-off rate (%)")
ax2.set_ylabel("Drop-off Rate (%)")

plt.title("DTP Dose 1 vs Dose 3 Coverage and Drop-off Rate")
fig.tight_layout()
plt.show()

print("Average DTP1 -> DTP3 drop-off rate across all years: {:.2f}%".format(
    dtp["DROPOFF_RATE_PCT"].mean()
))
print("\nMost recent 5 years:")
display(dtp.tail())
````

### Observation

The drop-off rate between the 1st and 3rd DTP dose has narrowed
considerably over time as health systems improved dose-completion
follow-up, but it has not reached zero — some children who receive
the first dose still do not receive the full series, which is exactly
the kind of gap the project's "Prioritize areas with low vaccination
coverage" use case is meant to surface.

## Question 3: How does vaccination coverage relate to disease
incidence?

We merge the Coverage dataset (MCV1 antigen) with the Incidence Rate
dataset (MEASLES) on country and year, then look at both the trend
over time and the country-level correlation.

````python
measles_coverage = coverage_df[coverage_df["ANTIGEN"] == "MCV1"][
    ["CODE", "NAME", "YEAR", "COVERAGE"]
].rename(columns={"COVERAGE": "MCV1_COVERAGE"})

measles_incidence = incidence_df[incidence_df["DISEASE"] == "MEASLES"][
    ["CODE", "YEAR", "INCIDENCE_RATE"]
].rename(columns={"INCIDENCE_RATE": "MEASLES_INCIDENCE_RATE"})

measles_merged = pd.merge(
    measles_coverage, measles_incidence,
    on=["CODE", "YEAR"], how="inner"
).dropna(subset=["MCV1_COVERAGE", "MEASLES_INCIDENCE_RATE"])

# --- Global trend over time ---
global_trend = measles_merged.groupby("YEAR")[
    ["MCV1_COVERAGE", "MEASLES_INCIDENCE_RATE"]
].mean().reset_index()

fig, ax1 = plt.subplots()
ax1.plot(global_trend["YEAR"], global_trend["MCV1_COVERAGE"],
         color="tab:blue", marker="o", label="MCV1 Coverage (%)")
ax1.set_ylabel("MCV1 Coverage (%)", color="tab:blue")
ax1.set_xlabel("Year")

ax2 = ax1.twinx()
ax2.plot(global_trend["YEAR"], global_trend["MEASLES_INCIDENCE_RATE"],
         color="tab:red", marker="o", label="Measles Incidence Rate")
ax2.set_ylabel("Measles Incidence Rate", color="tab:red")

plt.title("Measles Vaccination Coverage vs. Incidence Rate Over Time")
fig.tight_layout()
plt.show()

correlation = measles_merged["MCV1_COVERAGE"].corr(
    measles_merged["MEASLES_INCIDENCE_RATE"]
)
print(f"Correlation between MCV1 coverage and measles incidence rate: {correlation:.3f}")
````

### Observation

As MCV1 coverage rose globally, the measles incidence rate fell — the
correlation coefficient is negative, confirming the expected
relationship between vaccination coverage and disease control. This
directly supports the "Disease Prevention" use case: countries with
persistently high measles incidence despite reasonable coverage are
candidates for investigating vaccine effectiveness, cold-chain issues,
or under-reported coverage.

## Question 4: Which countries have high measles incidence despite
high vaccination coverage?

Using the most recent year available in the merged measles dataset, we
flag countries where MCV1 coverage is high (>= 80%) but the measles
incidence rate is still in the top 10 — a mismatch worth investigating.

````python
latest_year = measles_merged["YEAR"].max()
latest = measles_merged[measles_merged["YEAR"] == latest_year]

high_coverage_high_incidence = (
    latest[latest["MCV1_COVERAGE"] >= 80]
    .sort_values("MEASLES_INCIDENCE_RATE", ascending=False)
    .head(10)
)

print(f"Latest year in merged measles data: {latest_year}")
display(
    high_coverage_high_incidence[
        ["NAME", "MCV1_COVERAGE", "MEASLES_INCIDENCE_RATE"]
    ]
)

plt.figure()
sns.barplot(
    data=high_coverage_high_incidence,
    x="MEASLES_INCIDENCE_RATE", y="NAME",
    hue="NAME", legend=False, palette="Reds_r"
)
plt.title(f"Countries with High MCV1 Coverage (>=80%) but High Measles\nIncidence Rate — {latest_year}")
plt.xlabel("Measles Incidence Rate")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
````

### Observation

These are the countries where high reported coverage has not
translated into low disease incidence in the most recent year of
data. This is exactly the "Which regions have high disease incidence
despite high vaccination rates?" question from the project brief, and
flags candidates for deeper investigation into outbreak dynamics,
under-vaccinated pockets within the country, or data-quality issues
in the coverage estimate.

## Question 5: Are there disparities in vaccine introduction
timelines across WHO regions?

For each WHO region, we compute the average year in which each
vaccine was first introduced (`INTRODUCED_FLAG == "Yes"`), to see
which regions tend to adopt new vaccines earlier or later.

````python
introduced = vaccine_introduction_df[
    vaccine_introduction_df["INTRODUCED_FLAG"] == "Yes"
]

first_intro = (
    introduced.groupby(["WHO_REGION", "ISO_3_CODE", "DESCRIPTION"])["YEAR"]
    .min()
    .reset_index()
)

region_avg_intro_year = (
    first_intro.groupby("WHO_REGION")["YEAR"]
    .mean()
    .sort_values()
    .reset_index()
)

plt.figure()
sns.barplot(
    data=region_avg_intro_year, x="YEAR", y="WHO_REGION",
    hue="WHO_REGION", legend=False, palette="viridis"
)
plt.title("Average Year of First Vaccine Introduction by WHO Region")
plt.xlabel("Average Year of First Introduction")
plt.ylabel("WHO Region")
plt.xlim(region_avg_intro_year["YEAR"].min() - 2, region_avg_intro_year["YEAR"].max() + 2)
plt.tight_layout()
plt.show()

display(region_avg_intro_year)
````

### Observation

The average first-introduction year differs across WHO regions,
showing that some regions (often **EURO** and **AMRO**) tend to adopt
new vaccines earlier, on average, than others. This supports the
"Global Health Policy" use case of identifying regions that may need
support to introduce vaccines faster and close the adoption gap.

## Question 6: Which diseases have shown the largest reduction in
reported cases over time?

We compare total global reported cases in the earliest vs. the most
recent decade of data available for each disease.

````python
yearly_cases = (
    reported_cases_df.groupby(["DISEASE", "YEAR"])["CASES"]
    .sum()
    .reset_index()
)

early_years = yearly_cases[yearly_cases["YEAR"] <= yearly_cases["YEAR"].min() + 9]
recent_years = yearly_cases[yearly_cases["YEAR"] >= yearly_cases["YEAR"].max() - 9]

early_avg = early_years.groupby("DISEASE")["CASES"].mean().rename("EARLY_AVG_CASES")
recent_avg = recent_years.groupby("DISEASE")["CASES"].mean().rename("RECENT_AVG_CASES")

change = pd.concat([early_avg, recent_avg], axis=1).dropna()
change["PCT_CHANGE"] = (
    (change["RECENT_AVG_CASES"] - change["EARLY_AVG_CASES"])
    / change["EARLY_AVG_CASES"] * 100
)
change = change.sort_values("PCT_CHANGE")

plt.figure()
sns.barplot(
    data=change.reset_index(), x="PCT_CHANGE", y="DISEASE",
    hue="DISEASE", legend=False, palette="coolwarm"
)
plt.title("% Change in Average Reported Cases\n(First 10 Years vs. Last 10 Years of Data)")
plt.xlabel("% Change in Average Annual Cases")
plt.ylabel("Disease")
plt.tight_layout()
plt.show()

display(change)
````

### Observation

Diseases with a large negative percentage change (bars extending far
to the left) have seen the biggest reduction in reported cases
between the earliest and most recent decade of data — consistent with
the impact of sustained vaccination programmes. Diseases with a
positive change warrant closer investigation, since it may reflect
improved surveillance/reporting rather than an actual rise in cases.

As a final summary view, we build a correlation heatmap across the
